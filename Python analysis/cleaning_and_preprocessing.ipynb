```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)

df = pd.read_csv("healthcare_dataset.csv")

print("Dataset Shape:", df.shape)
df.head()
```

```python
for column in df.columns:
    print("\n", column)
    print(df[column].nunique())
```

```python
print("Gender:", df["Gender"].unique())
print("Medical Conditions:", df["Medical Condition"].unique())
print("Admission Types:", df["Admission Type"].unique())
print("Test Results:", df["Test Results"].unique())
print("Medications:", df["Medication"].unique())
```

```python
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"])

print(df[["Date of Admission", "Discharge Date"]].dtypes)
```

```python
print("Invalid ages:", ((df["Age"] < 0) | (df["Age"] > 120)).sum())

print("Negative billing:", (df["Billing Amount"] < 0).sum())
```

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
missing = df.isnull().sum()

print(missing)
```

```python
df.columns = [
    "Patient_Name",
    "Age",
    "Gender",
    "Blood_Type",
    "Medical_Condition",
    "Admission_Date",
    "Doctor",
    "Hospital",
    "Insurance_Provider",
    "Billing_Amount",
    "Room_Number",
    "Admission_Type",
    "Discharge_Date",
    "Medication",
    "Test_Results"
]

print(df.columns)
```

```python
df["Admission_Date"] = pd.to_datetime(
    df["Admission_Date"],
    errors="coerce"
)

df["Discharge_Date"] = pd.to_datetime(
    df["Discharge_Date"],
    errors="coerce"
)
```

```python
print(df[["Admission_Date", "Discharge_Date"]].dtypes)
```

```python
df["Length_of_Stay"] = (
    df["Discharge_Date"] - df["Admission_Date"]
).dt.days
```

```python
print(df["Length_of_Stay"].describe())
```

```python
print(
    "Invalid Length of Stay:",
    (df["Length_of_Stay"] < 0).sum()
)
```

```python
bins = [0, 18, 30, 45, 60, 75, 120]

labels = [
    "0-18",
    "19-30",
    "31-45",
    "46-60",
    "61-75",
    "76+"
]

df["Age_Group"] = pd.cut(
    df["Age"],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
df["Age_Group"].value_counts().sort_index()
```

```python
print("Invalid ages:", ((df["Age"] < 0) | (df["Age"] > 120)).sum())
```

```python
categorical_columns = [
    "Gender",
    "Blood_Type",
    "Medical_Condition",
    "Doctor",
    "Hospital",
    "Insurance_Provider",
    "Admission_Type",
    "Medication",
    "Test_Results"
]

for column in categorical_columns:
    df[column] = (
        df[column]
        .astype(str)
        .str.strip()
    )
```

```python
print("Gender:", df["Gender"].unique())

print("Medical Conditions:",
      df["Medical_Condition"].unique())

print("Admission Types:",
      df["Admission_Type"].unique())

print("Test Results:",
      df["Test_Results"].unique())
```

```python
df["Recovery_Flag"] = np.where(
    df["Test_Results"] == "Normal",
    1,
    0
)
```

```python
print(
    df["Recovery_Flag"].value_counts()
)
```

```python
recovery_rate = (
    df["Recovery_Flag"].mean() * 100
)

print(
    "Recovery Rate:",
    round(recovery_rate, 2),
    "%"
)
```

```python
df["Admission_Year"] = (
    df["Admission_Date"].dt.year
)

df["Admission_Month"] = (
    df["Admission_Date"].dt.month
)

df["Admission_Month_Name"] = (
    df["Admission_Date"].dt.strftime("%B")
)

df["Admission_Year_Month"] = (
    df["Admission_Date"]
    .dt.to_period("M")
    .astype(str)
)
```

```python
print(
    "Missing billing:",
    df["Billing_Amount"].isnull().sum()
)
```

```python
print(
    "Negative billing:",
    (df["Billing_Amount"] < 0).sum()
)
```

```python
print(
    df["Billing_Amount"].describe()
)
```

```python
print("Final shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:",
      df.duplicated().sum())

print("\nRecovery Flag:")
print(df["Recovery_Flag"].value_counts())
```

```python
df["Patient_Name"] = (
    df["Patient_Name"]
    .astype(str)
    .str.strip()
    .str.title()
)
print(df["Patient_Name"].head(20))
```

```python
print("Unique patient names:", df["Patient_Name"].nunique())
```

```python
negative_billing = df[df["Billing_Amount"] < 0]

print("Negative billing records:", len(negative_billing))
```

```python
negative_billing[[
    "Patient_Name",
    "Billing_Amount",
    "Medical_Condition",
    "Hospital"
]].head(10)
```

```python
negative_billing_percentage = (
    len(negative_billing) / len(df) * 100
)

print(
    f"Negative billing percentage: "
    f"{negative_billing_percentage:.2f}%"
)
```

```python
df = df[df["Billing_Amount"] >= 0].copy()

print("Records after removing invalid billing values:", len(df))
```
